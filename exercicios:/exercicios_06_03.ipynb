```python
#4 - Soma da Sequência
#Leia um número inteiro N positivo. Usando for, calcule e exiba a soma de todos os números de 1 até N. Valide que N é positivo.

num = int(input("Digite um número inteiro: "))

if num > 0:
    soma = 0
    for i in range(1, num + 1):
        soma+=1
        print(soma)

else:
    print("Número negativo")
```

```python
#5 - Contador de Pares
#Leia um número inteiro N positivo. Usando for e range(), exiba todos os números pares de 0 até N. Ao final, exiba a quantidade total de pares encontrados.

n = int(input("Digite um numero inteiro: "))
contador = 0

if n > 0:
    for i in range(0, n+1, 2):
      print(i)
      contador+=1
      print(f" tem {contador} pares")

else:
    print("Número negativo")
```

```python
#9 - Estrelas em Escada
#Leia um número inteiro N (1 a 10). Usando for com range() aninhado, exiba uma escada de asteriscos: a linha 1 tem 1 estrela, a linha 2 tem 2, até a linha N.

asteriscos = int(input("digite um valor de 1 a 10: "))

if 1<= asteriscos <=10:
    palavra = ''
    for i in range(asteriscos):
        palavra = palavra + '*'
        print(palavra)
else:
    print('ERROU!')
```

```python
#10 - Potências de Dois
#Usando for e range(), exiba as primeiras 10 potências de 2 (2⁰ a 2⁹), no formato '2^N = R'. Use o operador ** para calcular.

for i in range(10):
    resultado = 2 ** i
    print(f'2^ {i} = {resultado} ')
```

```python
#12 - Soma dos Ímpares
#Leia dois números inteiros A e B (A <= B). Usando for e range(), percorra o intervalo de A até B e calcule a soma de todos os números ímpares. Exiba a soma e a quantidade de ímpares encontrados.

A = int(input("Digite um numero inteiro: "))
B = int(input("Digite um numero inteiro: "))

somaI = 0
quantidade = 0

for i in range(A, B + 1):
    if i % 2 != 0:
        somaI += i
        quantidade += 1

print(f'soma dos impares: {somaI}')
print(f'quantidade de impares: {quantidade}')
```
