```python
#02 - Dias da Semana
#Leia um número de 1 a 7 e exiba o nome do dia correspondente (1 = Segunda, ..., 7 = Domingo). Para qualquer outro número, exiba "Número inválido".

dia = int(input("Digite um numero de 1 a 7 para representar o dia da semana: "))

match dia:
  case 1: print("Domingo")
  case 2: print("segunda-feira")
  case 3: print("terça-feira")
  case 4: print("quarta-feira")
  case 5: print("quinta-feira")
  case 6: print("sexta-feira")
  case 7: print("sábado")
  case _ : print("dia não válido")
```

```python
#7 - Leia dois números e um operador (+, -, *, /). Use elif para realizar a operação correta. Valide a divisão por zero. Para operador inválido, exiba "Operador desconhecido".

try:
    num1 = float(input("Digite um número: "))
    operador = input("Digite a operação (/*-+): ")
    num2 = float(input("Digite um número: "))

    if operador == '+':
        resultado = num1 + num2
        print(f"{num1} + {num2} = {resultado}")

    elif operador == '-':
        resultado = num1 - num2
        print(f"{num1} - {num2} = {resultado}")

    elif operador == '*':
        resultado = num1 * num2
        print(f"{num1} * {num2} = {resultado}")

    elif operador == '/':
      if num2 == 0:
        print("Erro: Divisão por zero!")
      else:
        resultado = num1 / num2
        print(f"{num1} / {num2} = {resultado}")

    else:
        print("Operador desconhecido")

except:
    print("Erro: Entrada inválida. Certifique-se de digitar números.")
```

```python
#10 - Desconto por Faixa de Compra
#Leia o valor de uma compra. Valide que é positivo. Aplique: ≥ R$500 → 15%, ≥ R$200 → 10%, ≥ R$100 → 5%, abaixo → 0%. Exiba o desconto e o valor final.

valor = float(input("Digite o valor de sua compra: "))

if valor >= 500:
  desconto = valor * 0.15
  vf = valor - desconto
  print(f"O valor de sua compra com o desconto de 15% sera de {vf}")

elif valor >= 200:
  desconto = valor * 0.10
  vf = valor - desconto
  print(f"O valor de sua compra com o desconto de 10% sera de {vf}")

elif valor >= 100:
  desconto = valor * 0.05
  vf = valor - desconto
  print(f"O valor de sua compra com o desconto de 5% sera de {vf}")

elif valor <100:
  vf = valor
  print(f"O valor de sua compra sera de {vf}")
```

```python
#13 - Tipo de Número
#Leia um número inteiro. Usando elif, classifique em: positivo e par, positivo e ímpar, negativo e par, negativo e ímpar, ou zero. Use o operador % para verificar paridade.

num = int(input("Digite um número inteiro: "))

if num == 0:
  print("Seu número é zero!")

elif num > 0 and num % 2 == 0:
  print("Seu número é positivo e par!")

elif num > 0 and num % 2 != 0:
  print("Seu número é positivo e impar!")

elif num < 0 and num % 2 == 0:
  print("Seu número é negativo e par!")

elif num < 0 and num % 2 != 0:
  print("Seu número é negativo e impar!")
```

```python
#14 - Aprovação com Frequência
#Leia a nota (0-10) e a frequência (0-100%) de um aluno. Use and para verificar aprovação: nota ≥ 7 e frequência ≥ 75%. Se apenas um critério falhar, indique qual. Se ambos falharem, exiba "Reprovado em nota e frequência".

nota = float(input("Digite a nota do aluno: "))
frequencia = float(input("Digite a frequencia do aluno(%): "))

if nota >= 7 and frequencia >= 75:
  print("aprovado!")

elif nota >= 7 and frequencia < 75:
  print("Reprovado em frequencia")

elif nota < 7 and frequencia >= 75:
  print("Reprovado em nota!")

elif nota < 7 and frequencia < 75:
  print("Reprovado em nota e frequencia!")
```
