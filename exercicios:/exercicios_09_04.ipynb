```python
senha_correta = "python2026"
tentativas = 0
max_tentativas = 3

while tentativas < max_tentativas:
    senha = input("Digite a senha: ")
    tentativas += 1

    if senha == senha_correta:
        print("Acesso liberado!")
        break

    else:
        restam = max_tentativas - tentativas
        if restam > 0:
            print(f"Senha incorreta! {restam} restante(s).")

if tentativas == max_tentativas:
    if senha != senha_correta:
        print("Conta bloqueada! Muitas tentativas")
```

```python
#Senha Segura
#Defina uma senha correta no código. Usando while, peça a senha ao usuário repetidamente até que ele acerte. A cada tentativa errada, exiba 'Senha incorreta, tente novamente.'. Ao acertar, exiba 'Acesso liberado!' e quantas tentativas foram necessárias.

senha_correta = "ifes2026"
contador = 0

while True:
    senha_digitada = input("Digite a senha: ")
    contador +=1

    if senha_digitada == senha_correta:
        print(f"Acesso liberado! Usou {contador} tentativas!")
        break

    else:
        print("Senha incorreta, tente novamente!")
```
