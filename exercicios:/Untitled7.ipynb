```python
#1 - LISTA SLIDE (LISTA):

notas = ['6' , '6.5' , '7' , '5' , '9' ]
print(f'Lista base:')

print(f'a) A primeira nota é: {notas[0]} e a última nota é: {notas[-1]}')

notas[1] = 10
print(f'b) {notas}')

notas.append(9)
print(f'c) {notas}')

print(f'd) Quantidade de notas: {len(notas)}')
```

```python
#2 - LISTA SLIDE (LISTA):

nomes = ['Ana' , 'João' , 'Maria' , 'Vitor']
#a)
print('Letra A')
for nome in nomes:
    print(f'Presente: {nome}')
nomes.append('Otávio')

#b)
print('\nletraB')
for nome in nomes:
    print(f'Presente: {nome}')
```

```python
#3 - LISTA SLIDE (LISTA):

numeros = ['1', '2' , '3' , '4' , '5' ,'6']

numeros.remove("3")
print(f'a){numeros}')

numeros.pop()
print(f'b) {numeros}')

numeros.sort()
print(f'c) {numeros}')

for i, valor in enumerate (numeros):
    print(f'{i+1}° nuúmero {valor}:')
```
