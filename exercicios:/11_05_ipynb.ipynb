```python
#3

numeros = (1 , 2 , 3 , 2 , 4 , 2 , 5)
contador = 0

for i in numeros:
    if numeros[i] == 2:
        contador += 1
        indice = numeros.index(2)

print(f'O primeiro número 2 aparece no índice: {indice} ')
print(f'O número 2 aparece {contador} vezes!')
```

```python
#LISTA SLIDE CONJUNTOS

#1
inscritos = ['Ana' , 'Bruno' , 'Ana' , 'Carla' , 'Bruno' , 'Diego']

print(set(inscritos))
print(len(inscritos))
```

```python
#2

turma_python = ['Ana' , 'Bruna' , 'Carla' , 'Diego']
turma_java = ['Carla' , 'Diego' , 'Elena' , 'Fabio']

turma_python = set(turma_python)
turma_java = set(turma_java)

#a) interseção:
print(turma_python & turma_java)

#b) união:
print(turma_python | turma_java)

#c) diferença:
print(turma_python - turma_java)
```
