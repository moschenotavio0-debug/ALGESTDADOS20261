```python
#01 - Validação de Limite Numérico
#Em um sistema de controle de acesso, é necessário verificar se um valor ultrapassa um limite definido.
#Elabore um programa em Python que solicite ao usuário a entrada de um número inteiro e determine se ele é maior que 10.
#Implemente a função verificar_maior_que_dez(numero), que receba um inteiro e retorne um valor booleano.
#O programa deve exibir uma mensagem descritiva ao usuário com base no resultado.

def verificar_maior_que_dez(numero):
    if numero > 10:
        return True
    else:
        return False

valor = int(input("Digite um número inteiro: "))

if verificar_maior_que_dez(valor):
    print("Maior que 10")

else:
    print("Igual ou menor que 10")
```

```python
#09 - Verificação de Ano Bissexto
#Em sistemas de calendário digital, a identificação de anos bissextos é essencial para o correto cômputo de datas.
#Crie um programa que leia um ano e informe se ele é bissexto.
#Implemente a função eh_bissexto(ano) com base na regra: divisível por 4, exceto centenas — salvo as divisíveis por 400.

def eh_bissexto(ano):
    if (ano%4 == 0) and (not(ano%100 == 0)) or (ano%400 == 0): return 'Bissexto'
    else: return 'Ano não bissexto'

ano = int(input("Digit o ano: "))
print(eh_bissexto(ano))
```

```python
#Parte II — Estruturas com Vetores e Funções · Questões 11–20
#11 - Soma de Elementos de um Vetor
#Em sistemas de análise estatística, é comum agregar valores armazenados em estruturas lineares.
#Crie a função soma_vetor(vetor) que receba uma lista de números inteiros e retorne a soma de todos os seus elementos, sem utilizar a função built-in sum().
#Teste com 10 valores fornecidos pelo usuário.

def soma_vetor(vetor):
    soma = 0
    for n in vetor: soma+=n
    return soma

numeros = [1,3,4,5,7,9,6,10,2]
print(soma_vetor(numeros))
```

```python
#12 - Ordenação Alfabética de Nomes
#Em um sistema de cadastro corporativo, nomes precisam ser organizados em ordem alfabética para facilitar a navegação.
#Implemente a função ordenar_nomes(nomes) que receba uma lista com 5 nomes e retorne uma nova lista ordenada alfabeticamente, sem modificar a lista original.
#Leia os nomes via entrada do usuário.

def ordenar_nomes(nomes):
    nomes.sort() #consertar depois
    return nomes

nomes = []
for nome in range(5):
    nome = input(f"Digite um nome: ")
    nomes.append(nome)

print(ordenar_nomes(nome))
```

```python
#13 - Identificação de Extremos em Conjunto de Dados
#Em análise de dados, determinar os valores extremos de um conjunto é operação essencial.
#Crie a função maior_menor(vetor) que receba uma lista de 8 números inteiros e retorne uma tupla no formato (maior, menor), sem utilizar as funções built-in max() e min().

def maior_menor(vetor):
    vetor.sort()
    return (vetor[-1],vetor[0])

lista = [8,5,2,3,6,9,7,4]
dados = maior_menor(lista)
print(dados)
```

```python
#14 - Filtragem de Números Pares
#Em um sistema de processamento de lotes, apenas valores com determinadas propriedades devem ser retidos.
#Implemente a função filtrar_pares(lista) que receba uma lista de inteiros e retorne uma nova lista contendo apenas os valores pares, preservando a ordem original.
#Teste com 10 números fornecidos pelo usuário.

def filtrar_pares(lista):
    pares = []
    for item in lista:
        if item%2==0:
            pares.append(item)
    return pares

lista = [8,5,2,3,6,9,7,4,1,10]
print(filtrar_pares(lista))
```
