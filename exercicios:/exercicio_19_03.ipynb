```python
#Número Par
#Solicite um número inteiro. Verifique se ele é par (divisível por 2) e exiba uma mensagem informando o resultado.

numero = int(input("digite um numero: "))

if numero % 2 == 0:
  print("O seu numero é um numero par!")

else:
  print("Seu número não é um número par!")
```

```python
#Temperatura Quente
#Leia a temperatura em graus Celsius. Se for maior que 30, exiba "Está quente hoje!". Caso contrário, exiba "Temperatura agradável.

temperatura = float(input("Digite a temperatura: "))

if temperatura > 30:
  print("Está quente hoje!")

else:
  print("Temperatura agradável!")
```

```python
#Saldo Suficiente
#Defina um saldo fixo (ex: 100.0). Peça ao usuário um valor para saque. Se o valor for menor ou igual ao saldo, exiba "Saque realizado!", caso contrário, "Saldo insuficiente."

saldo = 100
valor = float(input("Digite o valor de seu saque: "))

if valor <= saldo:
  print("Saque realizado!")

else:
  print("Saldo insuficiente!")
```

```python
#Número Zero
#Peça um número ao usuário. Exiba se o número é zero ou diferente de zero usando if/else.

numero = int(input("Digite um número:"))

if numero == 0:
  print("Seu número é zero!")

else:
  print("Seu número é diferente de zero!")
```

```python
#Nome Vazio
#Leia o nome do usuário. Se o nome for uma string vazia (falsy), exiba "Nenhum nome digitado.", caso contrário, exiba "Olá, [nome]!" usando f-string.

nome = input("Digite seu nome: ")

if nome == "":
  print("Nenhum nome digitado!")

else:
  print(f'Olá {nome}!')
```

```python
#Maior entre Dois
#Leia dois números inteiros. Exiba qual deles é o maior. Se forem iguais, exiba "Os números são iguais."

n1 = int(input("Digite um número inteiro: "))
n2 = int(input("Digite um número inteiro: "))

if n1 > n2:
  print("O seu primeiro número é maior!")

elif n1 == n2:
  print("Os dois números são iguais!")

else:
  print("O seu segundo número é maior!")
```
