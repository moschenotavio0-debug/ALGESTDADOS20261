```python
#02 - Produtos sem Repetição
#Leia cinco nomes de produtos digitados pelo usuário e armazene-os em um conjunto. Ao final, exiba todos os produtos cadastrados e a quantidade total de itens únicos.

conjunto = set()

for i in range(5):
    item = input(f"Digite o {i+1}° ")
    conjunto.add(item)

print()
for item in conjunto:
    print(item)

print(f"Produtos únicos:{(len(item))}")
```

```python
#03 - Analisando uma Tupla
#Crie uma tupla com 6 números inteiros. Exiba a quantidade de números pares, a soma dos números ímpares e o menor valor armazenado.

numeros = (1,2,3,4,5,6)
menor = numeros[0]
soma = 0
pares = 0
for numero in numeros:
    if numero%2 == 0:
        pares+=1

    else:
        soma+=numero

    if numero < menor:
        menor = numero

print(f'O menor valor é: {menor}\nA quantidade de pares é: {pares}\nA soma dos valores ímpares da tupla é {soma}.')
```

```python
#04 - União de Turmas
#Crie dois conjuntos contendo nomes de alunos de duas turmas diferentes. Exiba um novo conjunto contendo todos os alunos sem repetição usando a operação de união.

turmaA = ('a' , 'b' , 'c' , 'd' )
turmaB = ('e' , 'f' , 'g' , 'h')

turmaA = set(turmaA)
turmaB = set(turmaB)

print(turmaA | turmaB)
```

```python
#05 - Busca em Tupla
#Crie uma tupla com nomes de cidades. Leia uma cidade digitada pelo usuário e verifique se ela existe na tupla. Exiba uma mensagem informando se a cidade foi encontrada ou não.

cidades = ('Vila Pavão', 'Vilé Valéiro' , 'São Gbriel da Palha')
cidade = input("Digite o nome da cidade: ")

if cidade in cidades:
    print(f"A cidade de {cidade} foi encontrada!")

else:
    print(f"A cidade de {cidade} não foi encontrada")
```

```python
#06 - Conjunto de Vogais
#Leia uma palavra digitada pelo usuário e armazene em um conjunto todas as vogais diferentes encontradas nela. Exiba o conjunto final.

palavra = input("Digite uma palavra: ")
vogais = set()

for letra in palavra:
    if letra.lower() in 'aeiou':
        vogais.add(letra.lower())

print(vogais)
```

```python
#07 - Notas Imutáveis
#Crie uma tupla com quatro notas de um aluno. Exiba a média das notas, a maior nota e a menor nota. Mostre também todas as notas armazenadas.

notas = (2,4,6,8)
soma = 0
maior = max(notas)
menor = min(notas)

for nota in notas:
    soma += nota

media = soma / len(notas)

print(f"Média das notas é: {media}, a maior nota é: {maior}, a menor nota é: {menor}")
```

```python
#09 - Contador de Letras Únicas
#Leia uma frase e utilize um conjunto para armazenar apenas as letras diferentes digitadas pelo usuário. Ignore espaços. Ao final, exiba a quantidade de letras únicas encontradas.

frase = input("Digite uma frase: ").lower().replace(" ","")
letras = set()

for letra in frase:
    if letra not in letras:
        letras.add(letra)

print(letras)
```

```python
#10 - Tupla de Meses
#Crie uma tupla contendo os nomes dos 12 meses do ano. Leia um número de 1 a 12 e exiba o mês correspondente usando indexação da tupla.

meses = ('Janeiro','Fevereiro','Março','Abril','Maio','Junho','Julho','Agosto','Setembro','Outubro','Novembro','Dezembro')

numero = int(input('Digite um número de 1 a 12: '))

if 1 <= numero <= 12:
    print(meses[numero-1])

else:
    print('Mês não válido')
```
