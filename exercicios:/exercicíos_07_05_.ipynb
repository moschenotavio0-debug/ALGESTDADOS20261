```python
#1 - LISTA SLIDE (LISTA):

notas = ['6' , '6.5' , '7' , '5' , '9' ]
print(f'Lista base:')

print(f'a) A primeira nota é: {notas[0]} e a última nota é: {notas[-1]}')

notas[1] = 10
print(f'b) {notas}')

notas.append(9)
print(f'c) {notas}')

print(f'd) Quantidade de notas: {len(notas)}')
```

```python
#2 - LISTA SLIDE (LISTA):

nomes = ['Ana' , 'João' , 'Maria' , 'Vitor']
#a)
print('Letra A')
for nome in nomes:
    print(f'Presente: {nome}')
nomes.append('Otávio')

#b)
print('\nletraB')
for nome in nomes:
    print(f'Presente: {nome}')
```

```python
#3 - LISTA SLIDE (LISTA):

numeros = ['1', '2' , '3' , '4' , '5' ,'6']

numeros.remove("3")
print(f'a){numeros}')

numeros.pop()
print(f'b) {numeros}')

numeros.sort()
print(f'c) {numeros}')

for i, valor in enumerate (numeros):
    print(f'{i+1}° nuúmero {valor}:')
```

```python
#02 - Adição Dinâmica
#Inicie uma lista vazia chamada moedas_coletadas. Use um laço de repetição para pedir que o usuário informe 4 valores inteiros e adicione cada um deles ao final da lista usando append().

moedas_coletadas = []

for i in range(4):
   valor = int(input(f"Digite o {i+1}° valor: "))
   moedas_coletadas.append(valor)

print(moedas_coletadas)
```

```python
#03 - Inserção por Posição
#Crie a lista herois = ['Mario', 'Peach']. Utilize o método insert() para adicionar o nome 'Luigi' exatamente no meio da lista (índice 1) e exiba a lista final.

herois = ["Mário" , "Peach"]

herois.insert(len(herois)//2, "Luigi")
print(herois)
```

```python
#04 - Exclusão por Nome
#Dada a lista viloes = ['Bowser', 'Goomba', 'Wario', 'Koopa'], solicite que o usuário digite um nome. Se o nome estiver na lista, use remove() para excluí-lo.

viloes = ['Bowser', 'Goomba', 'Wario', 'Koopa']

nome = input("Digite um nome: ")

if nome in viloes:
    viloes.remove(nome)
    print(viloes)

else:
    print(viloes)
```

```python
#05 - Remover o Último Elemento
#Crie uma lista com nomes de 4 itens de poder (ex: Estrela, Cogumelo). Use pop() para remover o último item e mostre qual item foi removido e como a lista ficou.

powers = ["Estrela" , "Cogumelo Vermelho" , "Flor de Fogo" , "Pena"]
removido = powers[-1]
powers.pop()
print(f"Foi removido {removido} \ne a lista ficou: {powers}!")
```
