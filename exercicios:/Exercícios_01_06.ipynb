```python
#01 - Eliminação de Duplicatas com Conjuntos
#Em pipelines de processamento de dados, é comum receber listas com valores repetidos que precisam ser normalizados.
#Escreva um programa Python que receba uma lista de inteiros (definida diretamente no código, com ao menos 15 elementos e valores repetidos), converta-a para um set a fim de eliminar as duplicatas e, em seguida, converta o resultado de volta para uma list ordenada em ordem crescente.
#O programa deve exibir a lista original, o conjunto intermediário e a lista final, além de informar quantos elementos foram removidos no processo.

lista_inicial = [1,1,1,2,2,2,3,3,6,7,7,8,9]
conjunto =set(lista_inicial)
nova_lista = sorted(list(conjunto))

print(lista_inicial)
print(conjunto)
print(nova_lista)
print(f"foram removidos {len(lista_inicial)-len(nova_lista)} elementos")
```

```python
#02 - Construção de Dicionário a partir de Duas Listas
#A combinação de sequências paralelas em estruturas associativas é uma operação fundamental em manipulação de dados.
#Dado um par de listas — uma contendo nomes de produtos e outra contendo seus respectivos preços —, escreva um programa que utilize a função zip() para unir os elementos e construa um dict a partir dessa combinação.
#Em seguida, converta o dicionário resultante em uma lista de tuplas ordenadas pelo valor do preço (do menor para o maior) e exiba cada par no formato (produto, preço). Utilize ao menos 8 produtos no exemplo.

produtos = ['alho' , 'arroz', 'feijão', 'macarrão' , ' óleo']
precos = [15.99 , 21.90, 10.99, 5.99, 7.99]
dicionario =dict(zip(produtos,precos))
tuplas = tuple(sorted(dicionario.items(),key=lambda item: item[1]))
print(tuplas)
```

```python
#03 - Frequência de Elementos com Dicionário
#Contabilizar a ocorrência de elementos em uma sequência é uma tarefa recorrente em análise de dados e processamento de texto.
#Escreva um programa que receba uma lista de palavras (definida no código) e construa um dict que mapeie cada palavra à sua frequência de aparição na lista.
#Após construir o dicionário, converta-o em uma lista de tuplas (palavra, frequência), ordene-a pela frequência em ordem decrescente e exiba os cinco elementos mais frequentes.
#Por fim, extraia apenas as palavras (chaves) como um conjunto (set) e exiba o vocabulário único identificado.

palavras = ['abacate' , 'banana' , 'maçã' , 'uva', 'abacate' , 'banana' , 'maçã' , 'melancia']
dic_palavras = {}
for palavra in palavras:
    dic_palavras[palavra] = palavras.count(palavra)

print(dic_palavras)

tuplas = tuple(sorted(dic_palavras.items(),key=lambda item: item[1], reverse = True))


for i in range(5):
    print(tuplas[i])

final = set(dict(tuplas).keys())
print(final)
```
