```python
#REVISÃO TUPLAS: quae mesma coisa que lista, o que muda e que [] para ()

tupla = (0,1,3,3,5,7,7,7)
print(f"Conteudo da tupla: {tupla}")

qtde_elementos = len(tupla)
print(f"Quantidade de elemento na tupla: {qtde_elementos}")

qtde_setes = tupla.count(7)
print(f"Quantidade de 7 na tupla: {qtde_setes}")

elem_pos_4 =  tupla[4]
print(f"elemento na posição 4: {elem_pos_4}")

```

```python
#REVISÃO DE CONJUNTOS

conj1 = {1,2,3,4}
conj2 = {3,4,5,6}

print(f"O primeiro conjunto é {conj1}")
print(f"O segundo conjunto é {conj2}")

uniao = conj1.union(conj2)
interseccao = conj1.intersection(conj2)
diferenca = conj1.difference(conj2)

print(f'Uniao: {uniao}')
print(f'intersecçãp: {interseccao}')
print(f'diferença: {diferenca}')
```
