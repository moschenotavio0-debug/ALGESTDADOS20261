```python
#16 - Simulador de Bomba de Combustível
#Leia o tipo de combustível usando match/case ("gasolina", "etanol", "diesel") e a quantidade de litros. Valide que a quantidade é positiva. Cada combustível tem um preço por litro definido como variável no código. Com if aninhado, aplique 3% de desconto se o abastecimento for acima de 40 litros. Exiba o subtotal, o desconto (se houver) e o total a pagar.

combustivel = input("Digite qual o tipo de combustível (gasolina, etanol, diesel): ")
qtd_litros = float(input("Digite a qualtidade de litros: "))

if combustivel not in ["gasolina", "etanol", "diesel"]:
    print("O combustivel digitado está icorreto! Verifique e digite novamente!")

else:
    match combustivel:
        case "gasolina":
            valor = 6.70

        case "etanol":
            valor = 4.00

        case "diesel":
            valor = 7.50

    if qtd_litros <=0:
        print("A quantidade de litros deve ser positiva! Verifique o valor digitado e digite novamente!")

    else:
        if qtd_litros > 40:
            desconto = 0.03
            compl = "Desconto de 3%"

        else:
            desconto = 0
            compl = "Sem desconto"

        valor_subtotal = valor * qtd_litros
        valor_total = valor_subtotal - desconto

    print("")
    print(f"SUBTOTAL: R${valor_subtotal:.2f}")
    print(f"DESCONTO: {compl}")
    print(f"TOTAL A PAGAR: {valor_total:.2f}")

```

```python
#17 - Classificador de Velocidade
#Leia a velocidade em km/h e o tipo de via usando match/case ("urbana" → limite 60, "rodovia" → limite 100, "autopista" → limite 120). Valide que a velocidade não é negativa. Com if aninhado com elif, verifique: dentro do limite, excesso leve (até 20% acima), excesso moderado (até 50% acima) ou excesso grave (acima de 50%). Exiba o limite da via, a velocidade e a situação.

velocidade = float(input("Digite a velocidade (Km/h): "))
via = input("Digite o tipo de via (urbana, rodovia, autopista): ")

if via not in ["urbana", "rodovia", "autopista"]:
    print("O tipo de via está incorreto, confira e digite novamente!")

else:
    match via:
        case "urbana":
            limite = 60

        case "rodovia":
            limite = 100

        case "autopista":
            limite = 120

        case _:
            limite = None
            print("Erro: Tipo de via inválido.")

if velocidade <0 :
    print("A velocidade deve ser positiva!")

else:
    if limite is not None:
        print(f"Relatório de Velocidade")
        print(f"Via: {via} | Limite: {limite} km/h")
        print(f"Velocidade Registrada: {velocidade} km/h")

        if velocidade <= limite:
            situacao = "Dentro do limite"

        elif velocidade <= limite * 1.20:
            situacao = "Excesso Leve (Multa média)"

        elif velocidade <= limite * 1.50:
            situacao = "Excesso Moderado (Multa grave)"

        else:
            situacao = "Excesso Grave (Multa gravíssima + Suspensão)"

        print(f"Situação: {situacao}")
```

```python
#19 - Detector de Intervalo com Tolerância
#Leia um valor alvo, uma tolerância e um valor medido, todos como números positivos. Valide as entradas. Calcule os limites inferior (alvo - tolerância) e superior (alvo + tolerância). Use condicionais para classificar o valor medido em: dentro da faixa, abaixo do limite ou acima do limite. Calcule e exiba o desvio em relação ao alvo e a classificação.

alvo = float(input("Digite o valor alvo (positivo): "))
tolerancia = float(input("Digite a tolerância (positiva): "))
medido = float(input("Digite o valor medido (positivo): "))


if alvo <= 0 or tolerancia < 0 or medido <= 0:
    print("Erro: Todos os valores devem ser positivos (tolerância pode ser zero).")

    limite_inferior = alvo - tolerancia
    limite_superior = alvo + tolerancia

    desvio = medido - alvo

if limite_inferior <= medido <= limite_superior:
    classificacao = "DENTRO DA FAIXA"
elif medido < limite_inferior:
    classificacao = "ABAIXO DO LIMITE"
else:
    classificacao = "ACIMA DO LIMITE"

print("-" * 30)
print(f"Alvo: {alvo:.2f} (±{tolerancia:.2f})")
print(f"Limites: [{limite_inferior:.2f} - {limite_superior:.2f}]")
print(f"Medido: {medido:.2f}")
print(f"Desvio: {desvio:+.2f}")
print(f"Classificação: {classificacao}")
print("-" * 30)1
```

```python
#20 Simulador de Pedágio Inteligente
#Leia o tipo de veículo usando match/case ("moto", "carro", "caminhao", "onibus") e se possui tag de pagamento automático (s/n). Defina a tarifa base de cada veículo como variável. Com if aninhado, aplique 10% de desconto com tag. Leia também a hora atual, valide (0-23) e, com elif aninhado, aplique 20% de acréscimo em horário de pico (7-9 ou 17-19). Exiba a tarifa detalhada.

veiculo = input("Digite o tipo do seu veículo (moto, carro, caminhão ou ônibus): ")
tag = input("O seu veículo possui tag de pagamento automático (s/n)?: ")
horario = float(input("Digite o horário (00:00 - 23:00): "))

if horario < 0 or horario > 23:
    print("Horário inválido!")

else:
    match veiculo:
        case "moto":
            tarifa = 5.00

        case "carro":
            tarifa = 10.00

        case "onibus":
            tarifa = 15.00

        case "caminhão":
            tarifa = 20.00

        case "":
            tarifa = 0.00

    if tarifa is not None:
        if (7 <= horario <= 9) or (17 <= horario <= 19):
            adicional = tarifa * 0.20
            complemento_horario = "Horário de Pico (+20%)"

        else:
            adicional = 0
            complemento_horario = "Normal(sem adicional)"

        tarifa_atualizada = tarifa + adicional

        if tag == "s":
            desconto = tarifa_atualizada * 0.10
            complemento_tag = "Desconto de tag (-10%)"

        else:
            desconto = 0
            complemento_tag = "Sem desconto de tag!"

        tarifa_final = tarifa_atualizada - desconto

        print("")
        print(f"TAFIFA:")
        print(f"Veículo: {veiculo}")
        print(f"Tarifa base: R$:{tarifa:.2f}")
        print(f"Adicional Horário: R$:{tarifa_atualizada:.2f} - {complemento_horario}")
        print(f"VALOR FINAL: R$:{tarifa_final:.2f} - {complemento_tag}")

    else:
        print("Tipo de veículo inválido.")
```

```python
#28 - Simulador de Reserva de Assento
#Leia o número do assento (1 a 100) e a classe desejada usando match/case ("economica", "executiva", "primeira"). Valide o número. Com if aninhado, verifique se a faixa do assento corresponde à classe: 1-10 → primeira, 11-30 → executiva, 31-100 → econômica. Se incompatível, exiba mensagem. Se compatível, calcule o preço (econômica R$350, executiva R$890, primeira R$1.500) e exiba a confirmação.

print("-----CLASSES:-----")
print("Primeira Classe: assentos de 1 a 10")
print("Classe Executiva: assentos de 11 a 30")
print("Classe Econômica: assentos de 31 a 100")
print("")

assento = int(input("Digite o assento desejado (1 a 100): "))
classe = input("Digite a classe desejada (primeira, executiva ou econônico): ")

if assento < 0 or assento > 100:
    print("Assento incorreto! O assento deve ser selecioado entre 1 e 100!")

else:
    match classe:
        case "primeira":
            if 1 <= assento <= 10:
                print(f"Confirmado! Primeira Classe. Preço: R$1.500,00")

        case "executiva":
            if 11 <= assento <= 30:
                print(f"Confirmado! Classe Executiva. Preço: R$890,00")

        case "economica":
            if 31 <= assento <= 100:
                print(f"Confirmado! Classe Executiva. Preço: R$350,00")

        case _:
            print("Classe inválida!")

```

```python
#29 Simulador de Batalha por Tipos
#Leia o tipo do atacante e o tipo do defensor usando match/case. Tipos: "fogo", "agua", "planta", "eletrico", "terra". Relações de vantagem: fogo vence planta, água vence fogo, planta vence água, elétrico vence água, terra vence elétrico. Com if aninhado, determine se é vantagem (dano x2), desvantagem (dano x0.5) ou neutro (dano x1). Leia o poder base do ataque, calcule o dano final e exiba o resultado.

atacante = input("Digite o atacante (fogo, água, planta, eletrico ou terra): ")
defensor = input("Digite o defensor (fogo, água, planta, elétrico ou terra): ")
poder = float(input("Poder base do ataque: "))

multiplicador = 1.0

match atacante:
    case "fogo":
        if defensor == "planta": multiplicador = 2.0
        elif defensor == "agua": multiplicador = 0.5

    case "agua":
        if defensor == "fogo": multiplicador = 2.0
        elif defensor == "planta": multiplicador = 0.5
        elif defensor == "eletrico": multiplicador = 0.5

    case "planta":
        if defensor == "fogo": multiplicador = 0.5
        elif defensor == "agua": multiplicador = 2.0

    case "eletrico":
        if defensor == "agua": multiplicador = 2.0
        elif defensor == "terra": multiplicador = 0.5

    case "terra":
        if defensor == "eletrico": multiplicador = 0.5


dano_final = poder * multiplicador

if multiplicador == 2.0:
    print(f"É super efetivo! Dano final: {dano_final}")

elif multiplicador == 0.5:
    print(f"Não é muito efetivo... Dano final: {dano_final}")

else:
    print(f"Dano neutro. Dano final: {dano_final}")
```
