```python
#20 Simulador de Pedágio Inteligente
#Leia o tipo de veículo usando match/case ("moto", "carro", "caminhao", "onibus") e se possui tag de pagamento automático (s/n). Defina a tarifa base de cada veículo como variável. Com if aninhado, aplique 10% de desconto com tag. Leia também a hora atual, valide (0-23) e, com elif aninhado, aplique 20% de acréscimo em horário de pico (7-9 ou 17-19). Exiba a tarifa detalhada.

veiculo = input("Digite o tipo do seu veículo (moto, carro, caminhão ou ônibus): ")
tag = input("O seu veículo possui tag de pagamento automático (s/n)?: ")
horario = float(input("Digite o horário (00:00 - 23:00): "))

if horario < 0 or horario > 23:
    print("Horário inválido!")

else:
    match veiculo:
        case "moto":
            tarifa = 5.00

        case "carro":
            tarifa = 10.00

        case "onibus":
            tarifa = 15.00

        case "caminhão":
            tarifa = 20.00

        case "":
            tarifa = 0.00

    if tarifa is not None:
        if (7 <= horario <= 9) or (17 <= horario <= 19):
            adicional = tarifa * 0.20
            complemento_horario = "Horário de Pico (+20%)"

        else:
            adicional = 0
            complemento_horario = "Normal(sem adicional)"

        tarifa_atualizada = tarifa + adicional

        if tag == "s":
            desconto = tarifa_atualizada * 0.10
            complemento_tag = "Desconto de tag (-10%)"

        else:
            desconto = 0
            complemento_tag = "Sem desconto de tag!"

        tarifa_final = tarifa_atualizada - desconto

        print("")
        print(f"TAFIFA:")
        print(f"Veículo: {veiculo}")
        print(f"Tarifa base: R$:{tarifa:.2f}")
        print(f"Adicional Horário: R$:{tarifa_atualizada:.2f} - {complemento_horario}")
        print(f"VALOR FINAL: R$:{tarifa_final:.2f} - {complemento_tag}")

    else:
        print("Tipo de veículo inválido.")
```

```python
#24 Calculadora de Reajuste Salarial
#Leia o salário atual e o percentual de avaliação de desempenho (0 a 100). Valide ambos. Use elif para o reajuste: avaliação ≥ 90 → 15%, ≥ 75 → 10%, ≥ 60 → 5%, abaixo → 0%. Com if aninhado, verifique se o novo salário ultrapassa o teto de R$15.000 — se sim, ajuste ao teto e exiba aviso. Exiba salário anterior, percentual aplicado, reajuste em reais e novo salário.

salario_atual = float(input("Digite o valor de seu salário: "))
avaliação = int(input("Digite o percentual de avaliação de desempenho (0 a 100): "))


if avaliação < 0 or avaliação > 100:
    print("Erro! A avaliação deve estar entre 0 a 100!")

else:
    if avaliação >= 90:
        reajuste = salario_atual * 0.15
        salario_reajustado = salario_atual + reajuste

    elif avaliação >= 75:
        reajuste = salario_atual * 0.10
        salario_reajustado = salario_atual + reajuste

    elif avaliação >= 60:
        reajuste = salario_atual * 0.05
        salario_reajustado = salario_atual + reajuste

    else:
        reajuste = salario_atual + 0
        salario_reajustado = salario_atual + reajuste

aviso_teto = ""
if salario_reajustado > 15000:
    salario_reajustado = 15000
    aviso_teto ="(Ajustado ao teto de R$ 15.000,00)"

print("")
print("SALÁRIO:")
print(f"Salário anterior: R${salario_atual:.2f}")
print(f"O percentual aplicado foi de {avaliação}")
print(f"O reajuste, em R$, foi de: {reajuste:.2f}")
print(f"O salário atual é: R${salario_reajustado:.2f}")
```

```python
#25 Classificador de Triângulo com Ângulos
#Leia os três ângulos internos de um triângulo em graus. Valide que todos são positivos. Verifique se a soma é exatamente 180° — se não for, exiba "Triângulo inválido" e o valor da soma encontrada. Se válido, use elif para classificar quanto aos ângulos: acutângulo (todos < 90°), retângulo (algum == 90°) ou obtusângulo (algum > 90°). Com if aninhado, classifique também quanto aos lados com base nos ângulos iguais.

ang1 = float(input("Digite o angulo 1: "))
ang2 = float(input("Digite o angulo 2: "))
ang3 = float(input("Digite o angulo 3: "))
soma = ang1 + ang2 + ang3

if ang1 <= 0 or ang2 <= 0 or ang3 <= 0:
    print("Triângulo inválido! Todos os angulos devem ser positivos!")

else:
    if soma != 180:
        print(f"Triângulo invalido! O valor da soma foi de {soma}, e tem que ser 180°")

    else:
        if ang1 <90 and ang2<90 and ang3<90:
            tipo_angulo = "Octângulo"

        elif ang1==90 or ang2==90 or ang3 ==90:
            tipo_angulo = "Retângulo"

        else:
            tipo_angulo = "Obtusângulo"

    if ang1 == ang2 == ang3:
        tipo_lado = "Equilátero"

    elif ang1 == ang2 or ang1 == ang3 or ang2 == ang3:
        tipo_lado = "Isósceles"

    else:
        tipo_lado = "Escaleno"

    print("")
    print("RESULTADO:")
    print(f"Classificação: {tipo_angulo} e {tipo_lado}")
    print(f"Soma dos ângulos: {soma}°")
```
