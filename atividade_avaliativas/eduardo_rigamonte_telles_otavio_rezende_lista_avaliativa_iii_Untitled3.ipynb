```python
#01 - Crie uma lista com 15 números inteiros digitados pelo usuário.
#Em seguida, remova todos os números repetidos, mantendo a ordem original.

numeros = []

for i in range(15):
    num = int(input(f"Digite o {i+1}° número: "))
    numeros.append(num)

lista_sem_repeticao = []

for numero in numeros:
    if numero not in lista_sem_repeticao:
        lista_sem_repeticao.append(numero)

print(lista_sem_repeticao)
```

```python
#03- Faça um programa que leia uma lista de 10 números e divida todos os valores ímpares por 3 e os pares por 2, substituindo os valores originais.
 #(Use estruturas condicionais.)

numeros = []

for i in range(10):
    num = int(input("Digite um número: "))
    numeros.append(num)

for i in range(10):
    if numeros[i] % 2 == 0:
        numeros[i] = numeros[i] / 2
    else:
        numeros[i] = numeros[i] / 3

print(numeros)
```

```python
#05 - Leia 10 números inteiros e armazene-os em uma lista.
#Em seguida, crie uma segunda lista apenas com os números primos da primeira.

numeros = []
num_primos = []

for i in range(10):
    num = int(input("Digite um número: "))
    numeros.append(num)

for numero in numeros:

    if numero > 1:
        divisores = 0

        for i in range(1, numero + 1):
            if numero % i == 0:
                divisores += 1

        if divisores == 2:
            num_primos.append(numero)

print(num_primos)
```

```python
#06 - Escreva um programa que leia uma lista de 8 elementos e verifique (estrutura condicional) se ela é palíndroma (lê-se igual de trás para frente).

lista = []

for i in range(8):
    elemento = input("Digite um elemento: ")
    lista.append(elemento)

palindroma = True

for i in range(len(lista)):
    if lista[i] != lista[len(lista) - 1 - i]:
        palindroma = False

if palindroma:
    print("A lista é palíndroma.")
else:
    print("A lista não é palíndroma.")
```

```python
#Round III — Conjuntos (Sets)
#19 - Solicite dois conjuntos de 10 elementos fornecidos pelo usuário.
#Mostre a interseção, união e diferença (A − B e B − A).

conjuntoA = set()
conjuntoB = set()

for i in range(10):
    numero = int(input("Digite um número para o conjunto A: "))
    conjuntoA.add(numero)

for i in range(10):
    numero = int(input("Digite um número para o conjunto B: "))
    conjuntoB.add(numero)

print("União:")
print(conjuntoA | conjuntoB)

print("Interseção:")
print(conjuntoA & conjuntoB)

print("Diferença A - B:")
print(conjuntoA - conjuntoB)

print("Diferença B - A:")
print(conjuntoB - conjuntoA)
```

```python
#20 - Crie um conjunto com todas as letras usadas em uma frase digitada pelo usuário, excluindo espaços e pontuação.

frase = input("Digite uma frase: ")

letras = set()

pontuacoes = {" ", ".", ",", "!", "?", ";", ":", "(", ")", '"', "'"}

for caractere in frase:

    if caractere not in pontuacoes:
        letras.add(caractere)

print(letras)
```

```python
#21 - Escreva um programa que armazene em um conjunto todos os números entre 1 e 100 que não são múltiplos de 2 nem de 3.
#(Use estruturas de repetição e condicionais.)

numeros = set()

for i in range(1, 101):

    if i % 2 != 0 and i % 3 != 0:
        numeros.add(i)

print(numeros)
```

```python
#23 - Leia dois conjuntos de inteiros.
#Verifique (estrutura condicional) se um é subconjunto do outro.

conjA = set()
conjB = set()

for i in range(5):
    conjA.add(int(input("Digite um número para o conjunto A: ")))

for i in range(5):
    conjB.add(int(input("Digite um número para o conjunto B: ")))

subconjunto = True

for numero in conjuntoA:
    if numero not in conjuntoB:
        subconjunto = False

if subconjunto:
    print("A é subconjunto de B")
else:
    print("A não é subconjunto de B")
```

```python
#24 - Dado um conjunto de números inteiros, crie um novo conjunto com apenas os quadrados perfeitos encontrados.

numeros = set()

for i in range(10):
    numero = int(input("Digite um número: "))
    numeros.add(numero)

quadrados_perfeitos = set()

for numero in numeros:

    raiz = numero ** 0.5

    if raiz == int(raiz):
        quadrados_perfeitos.add(numero)

print(f"Os quadrados perfeitos são: {quadrados_perfeitos}")
```

```python
#Round IV — Dicionários
#25 - Crie um dicionário que armazene o nome e o salário de 5 funcionários.
#Calcule o reajuste de 10% para salários inferiores a R$ 2.000 e atualize o dicionário.
#(Use estruturas condicionais.)

funcionarios = {}

for i in range(3):
    nome = input("Nome do funcionário: ")
    salario = float(input("Salário: "))

    funcionarios[nome] = salario

for nome in funcionarios:
    if funcionarios[nome] < 2000:
        funcionarios[nome] = funcionarios[nome] * 1.10

print(funcionarios)
```

```python
#26 - Faça um programa que armazene em um dicionário o nome de 5 alunos e uma lista com suas 3 notas.
#Depois, exiba o nome e a média de cada aluno.

alunos = {}

for i in range(5):
    nome = input(f"Nome do {i+1}° aluno: ")

    notas = []

    for j in range(3):
        nota = float(input(f"Digite a {j+1} nota: "))
        notas.append(nota)

    alunos[nome] = notas

for nome in alunos:
    media = sum(alunos[nome]) / 3

    print(nome)
    print("Média:" , media)
```

```python
#28 - Solicite ao usuário 10 palavras e armazene em um dicionário onde a chave é a primeira letra e o valor é uma lista de palavras que começam com essa letra.

palavras = {}

for i in range(10):
    palavra = input("Digite uma palavra: ")

    letra = palavra[0]

    if letra not in palavras:
        palavras[letra] = []

    palavras[letra].append(palavra)

print(palavras)
```

```python
#29 - Crie um dicionário onde as chaves são nomes de produtos e os valores são listas com preços praticados em três lojas diferentes.
#Encontre o produto com maior variação de preço.

produtos = {}

for i in range(3):
    nome = input("Nome do produto: ")

    precos = []

    for j in range(3):
        preco = float(input("Preço: "))
        precos.append(preco)

    produtos[nome] = precos

maior_variacao = 0
produto_maior = ""

for produto in produtos:
    variacao = max(produtos[produto]) - min(produtos[produto])

    if variacao > maior_variacao:
        maior_variacao = variacao
        produto_maior = produto

print("Produto com maior variação:", produto_maior)
print("Variação:", maior_variacao)
```
