```python
#Round III — Conjuntos (Sets)

#19 - Solicite dois conjuntos de 10 elementos fornecidos pelo usuário.
#Mostre a interseção, união e diferença (A − B e B − A).

A = set()
B = set()

for i in range(10):
    A.add(input(f"Digite o {i+1}º elemento: "))

for i in range(10):
    B.add(input(f"Digite o {i+1}º elemento: "))

print("")
print("União:")
print(A | B)

print("")
print("Interseção:")
print(A & B)

print("")
print("Diferença A - B:")
print(A-B)

print("")
print("Diferença B - A:")
print(B-A)
```

```python
#20 - Crie um conjunto com todas as letras usadas em uma frase digitada pelo usuário, excluindo espaços e pontuação.

frase = input("Digite uma frase: ")
letras = set()
pontuacao  = [" ", ".", ",", "!", "?"]

for letra in frase:
    if letra not in pontuacao:
        letras.add(letra)

print(letras)
```

```python
#21 - Escreva um programa que armazene em um conjunto todos os números entre 1 e 100 que não são múltiplos de 2 nem de 3.
#(Use estruturas de repetição e condicionais.)

conjunto_numeros = set()

for i in range(1, 101):
    if i % 2 != 0 and i % 3 != 0:
        conjunto_numeros.add(i)

print("Números entre 1 e 100 não múltiplos de 2 nem de 3:")
print(conjunto_numeros)
```

```python
#22 - Solicite ao usuário que digite 10 palavras.
#Armazene em um conjunto todas as palavras distintas com mais de 5 letras.

palavras = set()

for i in range(10):
    palavra = input(f"Digite a {i+1}° palavra: ")

    if len(palavra) >5:
        palavras.add(palavra)

print(" ")
print("As palavras com mais de 5 letras são:")
print(palavras)
```

```python
#24 - Dado um conjunto de números inteiros, crie um novo conjunto com apenas os quadrados perfeitos encontrados.

num_inteiros = {1, 2, 4, 6,  9, 25, 30,  50}

quadrado_perfeito = set()

for numero in num_inteiros:
    raiz = numero ** 0.5

    if raiz == int(raiz):
        quadrado_perfeito.add(numero)

print(quadrado_perfeito)
```
